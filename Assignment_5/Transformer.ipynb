**Question 1.1 (5 points - Written):** Why doesn't RMSNorm need to compute the mean? In what way is "controlling scale" sufficient for gradient stability?

**Why RMSNorm doesn't need to compute the mean**: RMSNorm doesn’t need the mean because it only cares about keeping the overall size of the values under control, not shifting them up or down. It just scales things so they don’t get too big or too small.

**In what way "controlling scale" is sufficient for gradient stability**: Controlling the scale is enough because most training issues come from values blowing up or shrinking too much. By keeping their size stable, the model can learn more smoothly.

**Question 4.1 (3 points - Written):** Imagine you're deploying a chatbot that needs to handle
conversations of 50,000 tokens. Would you use standard attention or MHLA?
Why? What would be the memory savings?

**Which of standard attention or MHLA you would use**: MHLA

**Why**: Standard attention would use too much memory for a 50,000 token conversation, while MHLA keeps memory much smaller by compressing each token.

**What the memory savings would be**: The memory savings come from storing a smaller latent representation instead of full key and value vectors, which can reduce memory by several times depending on the dimensions used. This makes long conversations much more practical to handle.

**Question 4.2 (3 points - Written):** What's the main trade-off when using a smaller latent dimension
$d_{\text{latent}}$ in MHLA? How would you choose this hyperparameter?

**What the main trade-off when using a smaller latent dimension in MHLA is**: The main trade-off is between efficiency and information loss. A smaller latent dimension saves more memory but can lose important details, hurting performance. A larger latent dimension keeps more information but uses more memory.

**How you would choose the latent dimension**: You’d choose it by balancing performance and efficiency, usually starting with a moderate size and adjusting based on how much accuracy you lose versus how much memory you save.

**Question 4.3 (4 points - Written):** We used Pre-LN (normalize before the block) instead of Post-LN (normalize after the residual). Why is Pre-LN better for deep networks? Hint: think about gradient flow.

**Why Pre-LN is better for deep networks**: Pre-LN is better because it helps gradients flow more smoothly through the network during training. Since inputs are normalized before each layer, the model avoids very large or very small values that can make training unstable. This makes it easier to train deeper networks, while Post-LN can lead to gradients becoming unstable as depth increases.
